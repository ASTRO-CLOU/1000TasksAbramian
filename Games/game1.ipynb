```python
import random as r

plr_Ochki = 0
skynet_Ochki = 0
game_running = True

while game_running:
    skynet = r.randint(1, 3) 
    
    playerinput = input("\nВыберите камень, ножницы или бумага: ").upper()
    while playerinput not in ["КАМЕНЬ", "НОЖНИЦЫ", "БУМАГА"]:
        playerinput = input("Неправильный ввод! Введите камень, ножницы или бумага: ").upper()

    if skynet == 1:
        skynet = 'НОЖНИЦЫ'
    elif skynet == 2:
        skynet = 'КАМЕНЬ'
    else:
        skynet = 'БУМАГА'

    print(f"Игрок: {playerinput.capitalize()}\nСкайнет: {skynet.capitalize()}")

    if playerinput == skynet:
        print(f"Ничья\nИгрок: {plr_Ochki}, Скайнет: {skynet_Ochki}")
    elif (playerinput == 'НОЖНИЦЫ' and skynet == 'БУМАГА') or \
         (playerinput == 'КАМЕНЬ' and skynet == 'НОЖНИЦЫ') or \
         (playerinput == 'БУМАГА' and skynet == 'КАМЕНЬ'):
        plr_Ochki += 1
        print(f"Игрок победил в раунде!\nИгрок: {plr_Ochki}, Скайнет: {skynet_Ochki}")
    else:
        skynet_Ochki += 1
        print(f"Скайнет победил в раунде!\nИгрок: {plr_Ochki}, Скайнет: {skynet_Ochki}")

    if plr_Ochki == 3 or skynet_Ochki == 3:
        if plr_Ochki == 3:
            print("\nВы выиграли Скайнет!")
        else:
            print("\nВы проиграли, и Скайнет захватил мир...")

        while True:
            answer = input("Продолжить игру? Да или нет: ").upper()
            if answer == "ДА":
                plr_Ochki = 0
                skynet_Ochki = 0
                break
            elif answer == "НЕТ":
                game_running = False
                break 
            else:
                print("Надо ввести 'да' или 'нет'.")
```
